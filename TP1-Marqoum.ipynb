<h2>
  <span style="color:pink;">Objectif :</span>
</h2>
<ul style="color:black;">
  <li><strong>Variable floue :</strong> Distance_to_next_car</li>
  <li><strong>Univers de discours :</strong> [0, 30]</li>
  <li><strong>Ensembles flous :</strong>
    <ul>
      <li>v_small → très petite distance</li>
      <li>small → petite distance</li>
      <li>perfect → distance idéale</li>
      <li>big → grande distance</li>
      <li>v_big → très grande distance</li>
    </ul>
  </li>
</ul>


```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

# 1 Définition de l'univers de discours
x_distance = np.arange(0, 30.1, 0.1)

# 2 Définition des ensembles flous (trapézoïdaux et triangulaires)
v_small = fuzz.trapmf(x_distance, [0, 0, 2, 5])        # très petite distance
small   = fuzz.trimf(x_distance, [3, 6, 9])            # petite distance
perfect = fuzz.trimf(x_distance, [8, 12, 16])          # distance idéale
big     = fuzz.trimf(x_distance, [14, 18, 22])         # grande distance
v_big   = fuzz.trapmf(x_distance, [20, 25, 30, 30])    # très grande distance

# 3 Visualisation
plt.figure(figsize=(10,6))
plt.plot(x_distance, v_small, 'b', linewidth=2, label='v_small (très petite)')
plt.plot(x_distance, small, 'c', linewidth=2, label='small (petite)')
plt.plot(x_distance, perfect, 'g', linewidth=2, label='perfect (idéale)')
plt.plot(x_distance, big, 'orange', linewidth=2, label='big (grande)')
plt.plot(x_distance, v_big, 'r', linewidth=2, label='v_big (très grande)')

plt.title("Fonctions d'appartenance - Distance_to_next_car")
plt.xlabel("Distance (mètres)")
plt.ylabel("Degré d'appartenance")
plt.legend(loc="upper right")
plt.grid(True)
plt.show()
```

<h2>

</h2>
<ul style="color:black;">
  <li><strong>Variable floue :</strong> Speed_change</li>
  <li><strong>Univers de discours :</strong> [-20, 20]</li>
  <li><strong>Ensembles flous :</strong>
    <ul>
      <li>declining → vitesse en diminution</li>
      <li>constant → vitesse stable</li>
      <li>growing → vitesse en augmentation</li>
    </ul>
  </li>
</ul>


```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

# 1 Définition de l'univers de discours
x_speed_change = np.arange(-20, 20.1, 0.1)

# 2 Définition des ensembles flous
declining = fuzz.trapmf(x_speed_change, [-20, -20, -10, -2])   # vitesse en diminution
constant  = fuzz.trimf(x_speed_change, [-5, 0, 5])             # vitesse stable
growing   = fuzz.trapmf(x_speed_change, [2, 10, 20, 20])       # vitesse en augmentation

# 3 Visualisation
plt.figure(figsize=(10,6))
plt.plot(x_speed_change, declining, 'b', linewidth=2, label='declining (vitesse en diminution)')
plt.plot(x_speed_change, constant, 'g', linewidth=2, label='constant (vitesse stable)')
plt.plot(x_speed_change, growing, 'r', linewidth=2, label='growing (vitesse en augmentation)')

plt.title("Fonctions d'appartenance - Speed_change", fontsize=14)
plt.xlabel("Variation de la vitesse (km/h)", fontsize=12)
plt.ylabel("Degré d'appartenance", fontsize=12)
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

<h2>
  <span style="color:pink;">Objectif :</span>
</h2>
<ul style="color:black;">
  <li><strong>Variable floue de sortie :</strong> Acceleration_Adjust</li>
  <li><strong>Univers de discours :</strong> [-3, 3]</li>
  <li><strong>Ensembles flous :</strong>
    <ul>
      <li>neg_big → forte décélération</li>
      <li>neg_small → faible décélération</li>
      <li>zero → aucune variation</li>
      <li>pos_small → faible accélération</li>
      <li>pos_big → forte accélération</li>
    </ul>
  </li>
</ul>


```python
import numpy as np
import skfuzzy as fuzz
import matplotlib.pyplot as plt

# 1 Définition de l'univers de discours
x_accel = np.arange(-3, 3.1, 0.1)

# 2 Définition des ensembles flous (trapézoïdaux et triangulaires)
neg_big   = fuzz.trapmf(x_accel, [-3, -3, -2.5, -1.5])   # forte décélération
neg_small = fuzz.trimf(x_accel, [-2, -1, 0])             # faible décélération
zero      = fuzz.trimf(x_accel, [-0.5, 0, 0.5])          # aucune variation
pos_small = fuzz.trimf(x_accel, [0, 1, 2])               # faible accélération
pos_big   = fuzz.trapmf(x_accel, [1.5, 2.5, 3, 3])       # forte accélération

# 3 Visualisation
plt.figure(figsize=(10,6))
plt.plot(x_accel, neg_big, 'b', linewidth=2, label='neg_big (forte décélération)')
plt.plot(x_accel, neg_small, 'c', linewidth=2, label='neg_small (faible décélération)')
plt.plot(x_accel, zero, 'g', linewidth=2, label='zero (aucune variation)')
plt.plot(x_accel, pos_small, 'orange', linewidth=2, label='pos_small (faible accélération)')
plt.plot(x_accel, pos_big, 'r', linewidth=2, label='pos_big (forte accélération)')

plt.title("Fonctions d'appartenance - Acceleration_Adjust", fontsize=14)
plt.xlabel("Ajustement de l'accélération (m/s²)", fontsize=12)
plt.ylabel("Degré d'appartenance", fontsize=12)
plt.legend(loc="upper left")
plt.grid(True)
plt.show()
```

<h2>
  <span style="color:#C71585"> Exercice  4:</span> <span style="color:black;"></span>

<h2>
  <span style="color:pink;">Objectif :</span>
</h2>
<p style="color:black;">
Mettre en œuvre un système de régulation de vitesse automobile basé sur la logique floue, en modélisant les variables d’entrée et de sortie, en définissant une base de règles floues, et en simulant le comportement du véhicule pour ajuster automatiquement l’accélération.
</p>


```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
import warnings

# Supprimer les avertissements inutiles
warnings.filterwarnings("ignore")

# -----------------------------
# 1 Définition des variables floues
# -----------------------------
Distance_to_next_car = ctrl.Antecedent(np.arange(0, 30.1, 0.1), 'Distance_to_next_car')
Speed_change = ctrl.Antecedent(np.arange(-20, 20.1, 0.1), 'Speed_change')
Acceleration_Adjust = ctrl.Consequent(np.arange(-3, 3.1, 0.1), 'Acceleration_Adjust')

# -----------------------------
# 2 Fonctions d’appartenance
# -----------------------------
# Distance_to_next_car
Distance_to_next_car['v_small'] = fuzz.trapmf(Distance_to_next_car.universe, [0, 0, 2, 5])
Distance_to_next_car['small']   = fuzz.trimf(Distance_to_next_car.universe, [3, 6, 9])
Distance_to_next_car['perfect'] = fuzz.trimf(Distance_to_next_car.universe, [8, 12, 16])
Distance_to_next_car['big']     = fuzz.trimf(Distance_to_next_car.universe, [14, 18, 22])
Distance_to_next_car['v_big']   = fuzz.trapmf(Distance_to_next_car.universe, [20, 25, 30, 30])

# Speed_change
Speed_change['declining'] = fuzz.trapmf(Speed_change.universe, [-20, -20, -10, -2])
Speed_change['constant']  = fuzz.trimf(Speed_change.universe, [-5, 0, 5])
Speed_change['growing']   = fuzz.trapmf(Speed_change.universe, [2, 10, 20, 20])

# Acceleration_Adjust
Acceleration_Adjust['neg_big']   = fuzz.trapmf(Acceleration_Adjust.universe, [-3, -3, -2.5, -1.5])
Acceleration_Adjust['neg_small'] = fuzz.trimf(Acceleration_Adjust.universe, [-2, -1, 0])
Acceleration_Adjust['zero']      = fuzz.trimf(Acceleration_Adjust.universe, [-0.5, 0, 0.5])
Acceleration_Adjust['pos_small'] = fuzz.trimf(Acceleration_Adjust.universe, [0, 1, 2])
Acceleration_Adjust['pos_big']   = fuzz.trapmf(Acceleration_Adjust.universe, [1.5, 2.5, 3, 3])

# -----------------------------
# 3 Définition des règles floues
# -----------------------------
rules = [
    # v_small
    ctrl.Rule(Distance_to_next_car['v_small'] & Speed_change['declining'], Acceleration_Adjust['neg_small']),
    ctrl.Rule(Distance_to_next_car['v_small'] & Speed_change['constant'], Acceleration_Adjust['neg_big']),
    ctrl.Rule(Distance_to_next_car['v_small'] & Speed_change['growing'], Acceleration_Adjust['neg_big']),

    # small
    ctrl.Rule(Distance_to_next_car['small'] & Speed_change['declining'], Acceleration_Adjust['zero']),
    ctrl.Rule(Distance_to_next_car['small'] & Speed_change['constant'], Acceleration_Adjust['neg_small']),
    ctrl.Rule(Distance_to_next_car['small'] & Speed_change['growing'], Acceleration_Adjust['neg_big']),

    # perfect
    ctrl.Rule(Distance_to_next_car['perfect'] & Speed_change['declining'], Acceleration_Adjust['pos_small']),
    ctrl.Rule(Distance_to_next_car['perfect'] & Speed_change['constant'], Acceleration_Adjust['zero']),
    ctrl.Rule(Distance_to_next_car['perfect'] & Speed_change['growing'], Acceleration_Adjust['neg_small']),

    # big
    ctrl.Rule(Distance_to_next_car['big'] & Speed_change['declining'], Acceleration_Adjust['pos_big']),
    ctrl.Rule(Distance_to_next_car['big'] & Speed_change['constant'], Acceleration_Adjust['pos_small']),
    ctrl.Rule(Distance_to_next_car['big'] & Speed_change['growing'], Acceleration_Adjust['zero']),

    # v_big
    ctrl.Rule(Distance_to_next_car['v_big'] & Speed_change['declining'], Acceleration_Adjust['pos_big']),
    ctrl.Rule(Distance_to_next_car['v_big'] & Speed_change['constant'], Acceleration_Adjust['pos_big']),
    ctrl.Rule(Distance_to_next_car['v_big'] & Speed_change['growing'], Acceleration_Adjust['pos_small'])
]

# -----------------------------
# 4 Construction du système de contrôle flou
# -----------------------------
speed_ctrl_system = ctrl.ControlSystem(rules)
speed_simulation = ctrl.ControlSystemSimulation(speed_ctrl_system)

# -----------------------------
# 5 Simulation avec des valeurs d’entrée données
# -----------------------------
speed_simulation.input['Distance_to_next_car'] = 13   # m
speed_simulation.input['Speed_change'] = -2.5         # km/h

# Calcul de la sortie
speed_simulation.compute()

# -----------------------------
# 6 Affichage des résultats
# -----------------------------
print(" Distance_to_next_car = 13 m")
print(" Speed_change = -2.5 km/h")
print(f"  Acceleration_Adjust = {speed_simulation.output['Acceleration_Adjust']:.3f} m/s²")

# -----------------------------
# 7 Visualisation graphique
# -----------------------------
Acceleration_Adjust.view(sim=speed_simulation)
plt.show()
```
