Exercice 1 : Réseau sémantique et héritage

```python
# Définition du réseau
network_ex1 = {
    "Animal": {"properties": []},
    "Oiseau": {"isa": ["Animal"], "properties": ["peut_voler"]},
    "Mammifere": {"isa": ["Animal"], "properties": ["allaite"]},
    "ChauveSouris": {"isa": ["Oiseau", "Mammifere"], "properties": ["ne_pond_pas_d_oeufs"]}
}

# Fonction pour propager les propriétés
def propagate_properties(node, network):
    props = set(network[node].get("properties", []))
    for parent in network[node].get("isa", []):
        props.update(propagate_properties(parent, network))
    return props

print("Propriétés de la chauve-souris :", propagate_properties("ChauveSouris", network_ex1))
# Conflit : Chauve-souris hérite "peut_voler" mais ne pond pas d'œufs
```

Exercice 2 : Dauphins peuvent nager

Exercice 4 : Propagation d’activation

```python
network_ex4 = {
    "Chien": {"links": ["Animal", "Os"]},
    "Chat": {"links": ["Animal", "Lait"]},
    "Animal": {"links": ["EtreVivant"]},
    "Os": {"links": ["Nourriture"]},
    "Lait": {"links": []},
    "EtreVivant": {"links": []},
    "Nourriture": {"links": []}
}

def propagate_activation(node, level, factor):
    if level == 0:
        return {node: factor}
    result = {node: factor}
    for link in network_ex4[node]["links"]:
        result.update(propagate_activation(link, level-1, factor*0.5))
    return result

activation = propagate_activation("Chien", 3, 1.0)
print("Activation propagée depuis Chien :", activation)
```

Exercice 5 : Pingouin et Hibou

```python
network_ex5 = {
    "Oiseau": {"properties": ["peut_voler"]},
    "OiseauNocturne": {"isa": ["Oiseau"], "properties": ["chasse_la_nuit"]},
    "Hibou": {"isa": ["OiseauNocturne"], "properties": ["ne_vit_pas_dans_eau"]},
    "Pingouin": {"isa": ["Oiseau"], "properties": ["ne_peut_pas_voler"]}
}

def propagate_properties(node, network):
    props = set(network[node].get("properties", []))
    for parent in network[node].get("isa", []):
        props.update(propagate_properties(parent, network))
    return props

print("Propriétés héritées du Pingouin :", propagate_properties("Pingouin", network_ex5))
print("Propriétés héritées du Hibou :", propagate_properties("Hibou", network_ex5))
```

Exercice 6 : Traduction logique en réseau

```python
network_ex6 = {
    "Personne": {"properties": []},  # Ajouter cette ligne
    "Etudiant": {"isa": ["Personne"], "properties": ["non_travailleur"]},
    "Ali": {"isa": ["Etudiant"]}
}

def propagate_properties(node, network):
    props = set(network.get(node, {}).get("properties", []))
    for parent in network.get(node, {}).get("isa", []):
        props.update(propagate_properties(parent, network))
    return props

print("Propriétés d'Ali :", propagate_properties("Ali", network_ex6))
```

Exercice 7 : Réseau médical et inférence

```python
network_ex7 = {
    "MaladieRespiratoire": {"isa": ["Maladie"]},
    "Grippe": {"isa": ["MaladieRespiratoire"], "symptomes": ["Fièvre", "Toux"]},
    "Pneumonie": {"isa": ["MaladieRespiratoire"], "symptomes": ["Fièvre", "DouleurPoitrine"]}
}

patient_symptomes = ["Fièvre", "Toux"]

possible_maladies = []
for disease, info in network_ex7.items():
    if "symptomes" in info and all(s in patient_symptomes for s in info["symptomes"]):
        possible_maladies.append(disease)

print("Maladies possibles :", possible_maladies)
```
